```python
sentence_1 ="i had a great time at the movie it was really funny"
sentence_2 ="i had a great time at the movie but the parking was terrible"
sentence_3 ="i had a great time at the movie but the parking was not great"
sentence_4 ="i went to see a movie"
```

```python
# from textblob import TextBlob
%pip install textblob
```

```python
from textblob import TextBlob
print(sentence_1)
sentiment_score_1=TextBlob(sentence_1).sentiment.polarity
print("Sentiment Score:",sentiment_score_1)
```

```python
print(sentence_2)
sentiment_score_2 =TextBlob(sentence_2).sentiment.polarity
print("Sentiment Score:",sentiment_score_2)
```

```python
print(sentence_3)
sentiment_score_3 =TextBlob(sentence_3).sentiment.polarity
print("Sentiment Score:",sentiment_score_3)
```

```python
print(sentence_4)
sentiment_score_4 =TextBlob(sentence_4).sentiment.polarity
print("Sentiment Score:",sentiment_score_4)
```
