```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import spacy
import re
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bbc_data = pd.read_csv("bbc_news.csv")
bbc_data.head()
```

```python
bbc_data.info()
```

```python
title=pd.DataFrame(bbc_data['title'])
title.head()
```

cleaning the data

```python
#lowercase
title['lowercase']=title['title'].str.lower()
title.head()
```

```python
#stop word removal
en_stopwords = set(stopwords.words('english'))
title['stopword_removed'] = title['lowercase'].apply(lambda x: ' '.join([word for word in x.split() if word not in en_stopwords]))
title.head()
```

```python
#punctuation removal
title['no_stopword_punct_removed'] = title['stopword_removed'].apply(lambda x: re.sub(r'[^\w\s]', '', x))
title.head()
```

```python
#tokenize
title['token_raw']=title.apply(lambda x:word_tokenize(x['no_stopword_punct_removed']),axis=1)
title['tokens_clean'] =title.apply(lambda x:word_tokenize(x['no_stopword_punct_removed']),axis=1)
title.head()
```

```python
#lemmatizing
lemmatizer = WordNetLemmatizer()
title['token_clean_lemmatized'] = title['tokens_cleann'].apply(lambda x: [lemmatizer.lemmatize(word) for word in x])
title.head()
```

```python
token_raw_list =sum(title['token_raw'],[])
tokens_clean_list =sum(title['token_clean_lemmatized'],[])
```

post tagging

```python
nlp =spacy.load('en_core_web_sm')
```

```python
spacy_doc =nlp(' '.join(tokens_clean_list))
```

```python
pos_df=pd.DataFrame(columns=['token','pos_tag'])
for token in spacy_doc:
    pos_df =pd.concat([pos_df,pd.DataFrame.from_records({'token':[token.text],'pos_tag':[token.pos_]})],ignore_index=True)
```

```python
pos_df_count = pos_df.groupby(['token','pos_tag']).size().reset_index(name='counts').sort_values(by='counts',ascending=False)
pos_df_count.head(10)
```

```python
verbs =pos_df_count[pos_df_count.pos_tag=='VERB'][0:10]
verbs
```

```python
adj=pos_df_count[pos_df_count.pos_tag=='ADJ'][0:10 ]
adj
```
