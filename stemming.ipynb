```python
# study about stemming

import nltk
from nltk.stem import PorterStemmer
ps= PorterStemmer()
connect_tokens =['connect','connected','connecting','connection','connections','connectivity']
for i in connect_tokens:
    print(ps.stem(i))   
```

```python
learn_tokens =['learn','learned','learning','learner','learns']
for i in learn_tokens:
    print(i+':'+ ps.stem(i))   
```

```python
# lemmatization
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize('better',pos='a'))   
```

```python
lemmatizer =WordNetLemmatizer()
connect_tokens =['connect','connected','connecting','connection','connections','connectivity']
for i in connect_tokens:
    print(i+':'+ lemmatizer.lemmatize(i,pos='v'))
```
