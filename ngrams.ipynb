```python
# n gram

# n-grams help us to analyze the relationship between neighboring words
# n-gram =a sequence of n tokens
# n=1 unigrams
# n=2 =bigrams
#n=3 trigrams

```

```python
import nltk
import pandas as pd
import matplotlib.pyplot as plt
tokens =['the','rise','of','artiificial','intelligence','and','machine','learning   is','transforming','the','world of','technology and','business  and','creating','new','opportunities','and','challenges for','organizations and','individuals   alike.','the','integration of','ai and','ml is','driving innovation and','reshaping','industries across the globe.  from','healthcare to','finance,','education to','transportation,','ai and','ml are','revolutionizing the way we live and work.  as','we continue to','embrace the potential of ai and ml, it is crucial to','navigate the ethical considerations and ensure that these technologies are used responsibly for the benefit of society as a whole. ']
print(tokens)
```

```python
unigrams =(pd.Series(nltk.ngrams(tokens,1)).value_counts( ))
print(unigrams[:10])
```

```python
unigrams[:10].sort_values().plot.barh(color='lightsalmon',width=9,figsize=(12,0))
plt.title('ten most requently occuring unigrams')
```
