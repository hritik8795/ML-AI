```python
import re 
customer_review = ["saam was a great help to me in the store ","the casuier to to much rude about me ","amazing work from the seedan!","sarah was able to help me find the item i needed quickly ","lucky is such a great addition to a team "]
sarah_review =[]
pattern_to_find =r"sarah"
for string in customer_review:
    if re.search(pattern_to_find,string):
        sarah_review.append(string)
print(sarah_review)
```

```python
a_review =[]
pattern_to_find =r"^a"

for string in customer_review:
    if re.search(pattern_to_find,string):
        a_review.append(string)
print(a_review)
```

```python

# string end with y so use dollar
y_review =[]
pattern_to_find =r"y$"
for string in customer_review:
    if re.search(pattern_to_find,string):
        y_review.append(string)
print(y_review)
```

```python
needwant_review =[]
pattern_to_find =r"(need|want)"
for string in customer_review:
    if (re.search(pattern_to_find,string)):
        needwant_review.append(string)
print(needwant_review)
```

```python
# []:a set of characters we want to match ^="not",\w= word characters
no_punchuation_review =[]
pattern_to_find =r"[^\w\s]"
for string in customer_review:
    if re.sub(pattern_to_find,"",string):
        no_punchuation_review.append(string)
print(no_punchuation_review)    
```
